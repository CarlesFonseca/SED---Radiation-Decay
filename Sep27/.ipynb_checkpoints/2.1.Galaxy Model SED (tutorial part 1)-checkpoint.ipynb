We begin by analyzing Simple Stellar Population (SSP), which is the spectrum of a stellar population at one age and one metallicity. SSPs are grouped at the same metallicty for different ages and combined into a dat file. 

*For this example I've downloaded the file containing their fiducial, solar metallicity SSP models from version 2.2. The file is in the "data" folder and is called "spectra-bin-imf135_300.z020.dat".*

# Retrieve and Filter data (Step 1)

## Import raw format

```python
import numpy as np

raw_wavelengths = np.loadtxt("data/spectra-bin-imf135_300.z020.dat", usecols=0)
```

PERSONAL: We have the 'raw_grid' being a "large spreadsheet", where each: 
- Column represents a "snapshot" of the galaxy at a specific age
- Each row (the position of each row, its 'n array' position) is associated with a specific wavelength
- Individual cells contain the Flux for that specific wavelenth & age.

When the code runs raw_grid = np.loadtxt(... )[:, 1:], it is loading the entire table of data. The [:, 1:] part tells Python to ignore the first column (which is usually the wavelength list) and keep all the other columns—which are the flux values.

The author states that the nth column (where n is started at 1 with the wavelength column) is of age 10^(6+0.1*(n-2)) years. 

```python
ages = np.arange(2, 53)
ages = 10**(6+0.1*(ages-2))

raw_grid = np.loadtxt("data/spectra-bin-imf135_300.z020.dat")[:,1:] # load numerical data from a text file and immediately exclude the first column

print(raw_grid.shape)
```

As you've probably noticed, the raw grid is enormous (80 MB), so before we do anything else let's try to get it down to a manageable size. The easiest way of doing this is to interpolate the models onto a coarser grid in wavelength. We're also not going to be fitting any photometry that's bluer than 1000 Angstroms or redder than 6 microns, so we can get rid of regions outside this range entirely.

## Interpolate Data and Plot

Since we'll be creating a new wavelength array, we need to interpolate previous flux values to estimate the flux at our new wavelength. 

It iterates through each column, and it finds where each particular elements would sit between two points in the original 'raw_wavelengths'. Then, it computes a straight line b/t the fluxes at those old positions. By computing the value of that line at the new point it assigns the corresponding flux value. 

**Why Interpolation vs. Filtering?** (Grid Alignment) In astronomy, your observed data (from a telescope) almost never has the exact same wavelength "bins" as your theoretical models. If your model has a data point at $1000.5$ Å and your telescope measures at $1000.0$ Å, simple filtering won't help you compare them. Interpolation "stretches" or "squeezes" the model to perfectly match the target grid. 

In addition, Interpolation ensures your final data is perfectly spaced, which is a requirement for many mathematical algorithms (like Fast Fourier Transforms or certain Chi-squared fits).

```python
'''PROCEDURE: 
- Create np array between 1000 - 60000 armstrongs w/ steps of size = 10
- create an emtpy container (grid) filled with zeroes. 
        - height = length of new np array
        - width = number of cols in old grid
- interpolate 

Note: notation array[:, i] selects all ROWS from a specific column i
'''

wavelengths = np.arange(1000., 60000., 10.)

grid = np.zeros((wavelengths.shape[0], raw_grid.shape[1])) # rows of wavelength, cols of grid

for i in range(raw_grid.shape[1]):
    grid[:,i] = np.interp(wavelengths, raw_wavelengths, raw_grid[:,i])
    
print(grid.shape)
```

We also convert units of 'Solar luminosities per angstron for a star cluster with 10^6 Solar masses' to 'ergs/s/A per Solar mass of stars'

```python
grid /= 10**6
solar_luminosity = 3.827*10**33
grid *= solar_luminosity
```

```python
# We plot the model to check everything worked. 
import matplotlib.pyplot as plt

plt.figure(figsize=(20,12))
axes = [plt.subplot(311), plt.subplot(312), plt.subplot(313)]

'''
The author below plots wavelengths, as expected, and then grids. 

If we recall ages = 10**(6+0.1*(ages-2)), for each grid he's plotting the columns corresponding to: 
(i=0): 63 million years old
(i=1): 630 million years old
(i=2): 6.3 billion years old ---> each increasing i takes us further forward in cosmological history. 
'''
for i in range(3):
    ax = axes[i]
    ax.plot(wavelengths, grid[:,10*(i+2)]) 
    ax.set_xscale("log")
    ax.set_xlim(1000., 10000.)
    
plt.show()
```

We readily observe that the top plot (younger star) has lower wavelengths, corresponding to bluer stars. As time progresses, we see a reddening of the spectrum. 

# Telescope Filter Curves (Step 2)

In the first part of this notebook, we've created a plot for a given stellar population **model**. Since it's generated by a computer, we get a continuous, high-resolution line. Nevertheless, in real life, such an observation would be far too time-consuming (and, in turn, expensive). 

To fix this observation time issue, we instead apply filters to a telescope. Then, we don't see the whole spectrum at once and instead focus on a particular region. A **filter curve** represents how much light at a specific wavelength can pass through a telescope's filter to reach the detector. ---> the GitHub tutorial points out that filter curves can be downloaded from http://svo2.cab.inta-csic.es/svo/theory/fps/. 

What this second step of the notebook does is to predict what a model galaxy would look like through a specific filter (e.g., the HST F606W filter). To do that, it takes the model spectrum, multiplies it by the filter curve, and integrates (adds up) the result to get a single "synthetic" flux value to get the average flux across the entire range that the filter will receive light. 

## Setting Up the Filter Curves (create the models)

```python
import numpy as np

filter_names = ["data/filters/CFHT_u.txt",
                "data/filters/CFHT_g.txt",
                "data/filters/CFHT_r.txt",
                "data/filters/CFHT_i+i2.txt",
                "data/filters/CFHT_z.txt",
                "data/filters/subaru_z",
                "data/filters/VISTA_Y.txt",
                "data/filters/VISTA_J.txt",
                "data/filters/VISTA_H.txt",
                "data/filters/VISTA_Ks.txt",
                "data/filters/IRAC1",
                "data/filters/IRAC2"]

filter_curves = [] # array containing the array of values for each filter (array of arrays)

for name in filter_names:
    filter_curves.append(np.loadtxt(name))
```

```python
filter_curves # we see it has the left column wavelength (Earth frame) and right column filter value
```

All of these filter curves currently have transmission values at different sets of wavelength values. This is going to make it challenging to integrate our models over them. The best way of dealing with this is to resample your models and filters to the same wavelength values so you can just multiply them together and integrate. We've already resampled the models, so let's resample the filter curves to the same wavelengths.

NOTE: For a correct analysis this should be done with the data as is. Alternatively, we could show that the integrals are similar in both cases. wavelengths = np.arange(1000., 60000., 10.)

resampled_filter_curves = []

for filt in filter_curves:

    resampled_filter_curves.append(resampled_filt)


```python
wavelengths = np.arange(1000., 60000., 10.)

resampled_filter_curves = []

'''INTERPOLATION FUNCTION: 
np.interp(x, xp, fp), where
- x: The x-coordinates where you want to calculate interpolated values.
- xp: The x-coordinates of your known data points (must be increasing).
- fp: The y-coordinates (values) of your known data points.

Hence, the function below outputs the predicted values for the second column 
of the filters.txt (filter percentage value) at the desired evenly-spaced wavelengths. 
'''

for filt in filter_curves:
    resampled_filt = np.interp(wavelengths, filt[:,0], filt[:,1], left=0, right=0)
    resampled_filter_curves.append(resampled_filt)
```

```python
resampled_filter_curves
```

Now, we plot the model curves as in Step 1 of the tutorial. But this time, we include the filters. 

Nevertheless, the scales of fluxes are of 1e30, while filter values are between 0-1. To normalize, the tutorial does: 
- Divide each point in the grid (theoretical model plot) by the sum of all total values (hence, the area under the curve becomes 1).
- Similarly, he divides the filter transmission by its own sum, and then divides by a factor of 4 (arbitrary scaling factor for "better looks")

With this plot, we expect to verify alignment, checking:  
- Filters sit on the correct part of the spectrum
- The important features (breaks, peaks, etc) are covered by our filters (otherwise, we see flat data, which is useless to model)

```python
def get_model_grid():
    """ Loads up the BPASS grid of stellar models and 
    resamples it onto a coarser wavelength grid. (From Step 1)"""
    
    model_path = "data/spectra-bin-imf135_300.z020.dat"
    raw_wavelengths = np.loadtxt(model_path, usecols=0)
    raw_grid = np.loadtxt(model_path)[:,1:]
    
    grid = np.zeros((wavelengths.shape[0], raw_grid.shape[1]))

    for i in range(grid.shape[1]):
        grid[:,i] = np.interp(wavelengths, raw_wavelengths, raw_grid[:,i])
    
    grid *= (3.827*10**33)/(10**6)

    return grid


grid = get_model_grid()
```

```python
plt.figure(figsize=(20, 12))
axes = [plt.subplot(311), plt.subplot(312), plt.subplot(313)]

for i in range(3):
    ax = axes[i]
    
    ax.plot(wavelengths, grid[:,10*(i+2)]/grid[:,10*(i+2)].sum()) # normalizing (are under = 1)
    
    for j in range(12):
        normed_filt = resampled_filter_curves[j]
        normed_filt /= 4*resampled_filter_curves[j].sum() # arbitrary factor for scaling
        ax.plot(wavelengths, normed_filt)

    ax.set_xscale("log")
    ax.set_xlim(1000., 60000.)
    
plt.show()
```

Author's note: Great, now we have our models and our filters on the same wavelength sampling, so all we need to do is to integrate over the product of the two to get our fluxes. Simple. Hmm... except that our observed galaxies aren't going to be in the rest-frame. Which means that our spectrum is going to shift to the right whilst our filter curves stay put, which is going to mess everything up. Damn.

## Accounting for Redshift

There are probably as many different ways of dealing with this problem as there are SED fitting codes. The obvious way is simply to redshift the spectral model each time you change redshift and then resample it back onto the 10 Angstron wavelength grid. However in my experience, the grids of stellar models are usually far more cumbersome than the filter curve files, so an alternative and faster method is to blueshift the filters each time you change redshift. We can write a quick function to do this like so.

PERSONAL: We know the simplified form for redshift is: $$z=\frac{\lambda_{observed} - \lambda_{rest}}{\lambda_{rest}} = \frac{\lambda_{observed}}{\lambda_{rest}} -1$$

So simplifying we get: 
$$\lambda_{rest} = \frac{\lambda_{observed}}{1+z} $$ 

By applying this formula to our filters, we're "blue-shifting" them, meaning that we're analyzing what their corresponding wavelengths would have been if no redshift were present. 

**Redshift works different for high redsfhit ---> Use integral formula.**

```python
def blueshift_filters(redshift):
    """ A function that resamples filters onto the same wavelength
    basis as the model spectrum at the specified redshift """
        
    resampled_filter_curves = []

    for filt in filter_curves:
        blueshifted_wavs = filt[:, 0]/(1 + redshift) 
        
        resampled_filt = np.interp(wavelengths,
                                   blueshifted_wavs, filt[:, 1],
                                   left=0, right=0)
        
        resampled_filter_curves.append(resampled_filt)
        
    return resampled_filter_curves
```

Let's make a quick test plot to see if this works / makes sense: 

```python
redshift = 1.5

resampled_filter_curves = blueshift_filters(redshift)

plt.figure(figsize=(20, 12))
axes = [plt.subplot(311), plt.subplot(312), plt.subplot(313)]

redshifted_wavs = wavelengths*(1+redshift)

for i in range(3):
    ax = axes[i]
    
    ax.plot(redshifted_wavs, grid[:,10*(i + 2)]/grid[:,10*(i + 2)].sum())
    
    for j in range(12):
        normed_filt = resampled_filter_curves[j]
        normed_filt /= 10*resampled_filter_curves[j].sum()
        ax.plot(redshifted_wavs, normed_filt)

    ax.set_xscale("log")
    ax.set_xlim(1000.*(1 + redshift), 60000.*(1 + redshift))
    
plt.show()
```

Great, now we can generate filter curves that will sample our spectrum as if it were placed at any redshift we like. We still have to deal with converting our spectra from luminosities to fluxes at the observed redshift, but we'll cover that in the next example.

# From Luminosity to Flux (Step 3)

Now we can think about how we're going to generate the photometry we want to fit to our observational data for a given redshift and age. All we really need to do is convert the luminosity units (erg/s/A per Solar mass of stars) into flux units (erg/s/cm^2/A per Solar mass of stars) at the observed redshift and integrate over the filter curves.

## Theoretical Approach (Carles & Gemini explanation)
On a telescope, we might receive some light, but we know that's not going to be all that was emitted from the galaxy. We need to account for: 
- Redshifting of Light as it is emitted (receding galaxy) and as light traverses the cosmos
- Percentage of Area that we occupy in a luminosity sphere (since luminosity is emitted equally in all directions).

We define $L_{bolometric}$ (Luminosity) to be the total **power output** of the object, measured in units of Energy / Time. The subscript 'bolometric' denotes that it is summed over all wavelengths (radio, UV, infrared, Optical, X-Ray summed up all together). 

To account for the portion of the spherical shell we occpy, we define $F_{bolometric}$ (Flux) to be the **Power Density** reaching our detector. Its units are in Flux: Energy / (Time $\times$ Area). 

Finally, taking into account the redshift factor, we obtain: $$F_{bolometric} = \frac{L_{bolometric}}{4 \pi d_L^2}$$

Here, $d_L$ (also luminosity distance) is the "effective" distance, and NOT not the physical distance you would measure with a ruler today (that is proper distance). Instead, $d_L = d_{comoving} \times (1+z)$, where the '1+z' factor expresses the cosmological effects of universe expansion. 


Finally, it is important to note that we are not concerned only with bolometric Flux, but rather with flux at a given wavelength. Since the x-axis has stretched (because of redshift effects), the same amount of light is now spread over a wider wavelength interval (Amount of Light per Angstrom has dropped). To correct for this effect, we must include a $\times \frac{1}{1+z}$ factor to our previous computations. In conclusion, we get: $$F_\lambda = \frac{L_\lambda}{4\pi d_L^2} \times \frac{1}{(1+z)}$$ The subscript 'lambdas' in this formula indicate that it's not bolometric measurements, but rather at a given wavelength (and hence, the extra $\times \frac{1}{1+z}$ factor). This will come into play later in this Section for Photometric Fluxes. 

## Application of Flux Conversion Formulas
In this notebook, the author does: To convert to flux units at the observed redshifts we need to divide our luminosity by 4piluminosity_distance(z)^2, and then divide by (1+redshift) to account for the spreading out of the light. We can calculate the luminosity distance using the astropy.cosmology module, we'll use redshift 1 for this example and take a random SSP from our model grid.

```python
from astropy.cosmology import FlatLambdaCDM

cosmo = FlatLambdaCDM(H0=70., Om0=0.3)

ssp_model = grid[:,25] # Selecting column 25!!!

redshift = 1 # arbitrary value which can be changed!!
 
luminosity_distance = cosmo.luminosity_distance(redshift).value # (in megaparsecs)


# Convert to centimetres
cm_per_Mpc = 3.086*10**24
luminosity_distance *= cm_per_Mpc

# Spread out the total flux emitted across a sphere 
# with the luminosity distance as its radius.
ssp_model /= 4*np.pi*luminosity_distance**2

# Account for the fact the same amount of light is
# now spread out over a larger wavelength interval.
ssp_model /= (1 + redshift)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 5))
plt.plot(wavelengths*(1 + redshift), ssp_model)
plt.xscale("log")
plt.xlim(1000.*(1 + redshift), 60000.*(1 + redshift))
plt.show()
```

We notice this graph looks quite different from previous ones. It is important to note that the axis has been shifted to higher wavelengths (from redshifted wavelengths to rest ones). Moreover, the y-axis has changed in values and magnitude because we are now plotting flux rather than luminosity. 

What we have done, is transform the intrinsic luminosity of our theoretical model spectrum ($L_\lambda$) into Observed Flux Density ($F_\lambda$) applying the cosmological math. IMPORTANT: The SPP model contains the flux that each rest-frame wavelength will contribute to Earth. Yet, we shift the x-axis to account for the stretching of wavelength. 

In real cases (e.g. HST), however, we obtain "counts" of sensors which can be converted to Energy from callibration against known objects. Given that the counts are measured in a time interval and through a telescope (and filter) with a given width, we end up getting Flux per angstrom. We then reverse the formula to get the intrinsic luminosity of our source. 

## Photometric Fluxes (Filter Integration)
Above we have the spectrum in erg/s/cm^2/A per Solar mass of stars. Now let's turn this spectrum into photometric fluxes by integrating over our filters.

Recap of what we had: 
- spp model: fluxes (that we will receive on Earth) associated to each rest-frame (source) wavelength
- blueshift_filters(redshift): filter values at rest-frame (source) wavelengths

We now want to find the individual contribution of each filter to flux. To do so, we multiply ssp * filter_curves_z, killing all light except for the part that falls inside that specific filter's window. 

Then, we perform the integration for the flux_contributions and redshifted_wavs. We use redshifted waves because we compute an integral of flux over wavelength, which has been stretched out due to z. In Section 'Theoretical Approach (Carles & Gemini explanation)', we already discussed how the $F_{\lambda}$ value had an extra $\times \frac{1}{1+z}$ factor, because of the compressed nature of the rest-frame wavelength. In this final step, we are merely stretching that wavelength again and computing the integral. 

Finally, we must normalize the integral of computed flux from a filter by the length of the filter curve over that wavelength to obtain the **Mean Flux Density** (to prevent larger "buckets" of filter area from outshining other significant features). We are therefore obtaining the average height of the spectrum within that filter's window. 

```python
filter_curves_z = blueshift_filters(redshift)

photometry = np.zeros(len(filter_curves))

redshifted_wavs = wavelengths*(1 + redshift)

'''ALGORITHM
Obtain flux_contributions by taking the blueshifted filters with the spps_model. In this way, both wavelength axes are similar. 
'''

for i in range(photometry.shape[0]):
    flux_contributions = filter_curves_z[i]*ssp_model
    photometry[i] = np.trapezoid(flux_contributions, x=redshifted_wavs)
    photometry[i] /= np.trapezoid(filter_curves_z[i], x=redshifted_wavs)
    
print(photometry)
```

If we with to plot these photometries to check they agree with our theoretical curve (ssp model, to get what we would observe on Earth), we must first know what the corresponding x-value would be for each point. To get this, we multiply each wavelength by its corresponding filtering probabilistic weight.

That is, to get the weighted average of wavelength with respect to transmission, we do: w_average = sum(T*w) / sum(T), where T is the probabilistic transmission of the filter. In the code, we will have: 
- wavelength = filt[:, 0]
- Transmission = filt[:, 1]

```python
eff_wavs = np.zeros(len(filter_curves)) # number of filter curves we have

# Calculate the effective wavelengths of the filter curves
for i in range(len(filter_curves)):
    filt = filter_curves[i]

    wav_weights = filt[:,1]*filt[:,0] # weight of each wavelength in x-position computation
    
    eff_wavs[i] = np.trapezoid(wav_weights, x=filt[:,0]) # integral of wave_weights over the Earth frame wavelengths

    eff_wavs[i] /= np.trapezoid(filt[:, 1], x=filt[:,0]) # normalize (divide by area of probabilistic transmission curve)

# Plot the model photometry over the model spectrum
plt.figure(figsize=(15, 5))
plt.plot(wavelengths*(1 + redshift), ssp_model)
plt.scatter(eff_wavs, photometry, color="red", s=250, zorder=10)
plt.xscale("log")
plt.xlim(1000.*(1 + redshift), 60000.*(1 + redshift))
plt.show()
```
