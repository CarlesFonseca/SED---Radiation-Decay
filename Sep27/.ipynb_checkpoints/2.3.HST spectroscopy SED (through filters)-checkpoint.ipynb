# Retrieving Hubble Data - Preliminary Analysis

## Discovery Portal: 

There is also the possibility of going to the MAST Discovery Portal: http://mast.stsci.edu/ , where we can search an object (3C273) and filter by RA, DEC, instrument, etc. 

If we try to download a given object from there (in our case, 'hst_13764_01_acs_wfc_f606w-pol0v_jcm101'), we retrieve 3 files. 

### Perplexity explanation:
We retrieve science image plus two catalog products derived from that image. The FITS file is the actual calibrated HST image; the point-cat.ecsv and segment-cat.ecsv files are catalog tables generated from source detection and segmentation on that image, which is why you do not get only one file.
- hst_13764_01_acs_wfc_f606w-pol0v_jcm101_drz.fits: this is the main drizzled science image in FITS format, using ACS/WFC in F606W with polarization-related metadata in the filename. The drz suffix indicates a pipeline-combined, geometrically corrected image product.
- point-cat.ecsv: this is a source catalog of detected point-like objects in the image, typically with positions, fluxes, and quality measurements.
- segment-cat.ecsv: this is a segmentation catalog, meaning the image has been divided into labeled regions or “segments” associated with detected structures, often useful for extended-source analysis.

### Why there are multiple files
HST archive products are often delivered as a multi-extension FITS image plus separate derived catalogs, because the image itself and the source-detection results serve different purposes. The FITS file contains the pixel data and metadata, while the ECSV catalogs are higher-level products made by a pipeline or archive service to help users identify objects or regions in the image. For jet work, the FITS image is usually the primary file you want; the catalog files are optional unless you want archive-detected sources or region masks.

## Examining Downloaded Files: 

### File Structure

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from astropy.io import fits
```

```python
hdul = fits.open("hst_13764_01_acs_wfc_f606w-pol0v_jcm101/hst_13764_01_acs_wfc_f606w-pol0v_jcm101_drz.fits")
hdul.info()
```

Your FITS file contains the standard HST drizzled image structure plus an associated header table. In your case:

    PRIMARY: primary header only, with observation metadata.

    SCI: the drizzled science image, which is the main array you will analyze.

    WHT: the weight map, telling you the effective inverse-variance or exposure weight per pixel; this is useful for masking and error estimation.

    CTX: the context image, showing which input pixels contributed to each output pixel during drizzling.

    HDRTAB: a binary table containing per-exposure header metadata for the observations contributing to the combined product.

Which one to focus on? For your science analysis, focus on:

    SCI for the image itself.

    WHT for uncertainties, masks, and low-weight regions.

    HDRTAB if you need to inspect the individual input exposures and their metadata.

- Use the SCI image to define the jet region.
- Use the WHT map to reject poorly constrained pixels.
- Verify from the observation metadata whether the exposure was intended for polarimetry.
- If it was, be cautious about using the flux as a standard broadband SED point because the polarizer changes the effective throughput.

From this header, we confirm this fits file is the HST observation after correction:

We observe most of the 'CALIBRATION SWITCHES' have been completed. Also, those that have not been completed are markes as OMIT, rather than PERFORM, so we assume there's no more necessary post-processing on that end.

### Important Parameters
For our initial analysis, we're interested in the paremeters: 
- RA_TARG =   1.872779154167E+02 / right ascension of the target (deg) (J2000)
- DEC_TARG=   2.052388888889E+00 / declination of the target (deg) (J2000)
- EXPTIME =               7612.0 / exposure duration (seconds)--calculated    *(There's also date observed if desired)*
- OBSTYPE = 'IMAGING       '     / observation type - imaging or spectroscopic
- DETECTOR= 'WFC'                / detector in use: WFC, HRC, or SBC
- FILTER1 = 'F606W             ' / element selected from filter wheel 1
- FILTER2 = 'POL0V             ' / element selected from filter wheel 2


```python
nan_mask = ~np.isnan(hdul[1].data)
data_masked = hdul[1].data[nan_mask]

print(np.sum(data_masked))
num_points = len(data_masked)
x_array = np.linspace(0, 1, num = num_points)

plt.scatter(x_array, data_masked, s=0.1, alpha = 0.2)
plt.show()
```

This snippet of code is run to indeed check that we have values and not merely 'nan' correpsonding to no usable observations.

In the above case, we see clearly that we need to select HST $\rightarrow$ ACS_WFC $\rightarrow$ F606W. This corresponds to file: HST/ACS_WFC.F606W (https://svo2.cab.inta-csic.es/svo/theory/fps/index.php?id=HST/ACS_WFC.F606W&&mode=browse&gname=HST&gname2=ACS_WFC#filter)
